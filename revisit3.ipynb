###  tax application on salary per annum 5% tax

```python
taxable_salary = 100000
spa = float(input("ENTER SPA"))
if spa > taxable_salary:
    tax_amt = spa*0.05
    net_sal = spa - tax_amt
    print(f"your spa is: {spa}")
    print(f"tax apply on your spa is: {tax_amt}")
    print(f"your net salary becomes:{net_sal}")
else:
    print("no tax")
```

```python
salary = int(input("Enter your Salary:"))
if salary >= 240000:
    tax_amount = salary*.20
elif salary >= 120000:
    tax_amount = salary*.15
elif salary >= 90000:
    tax_amount = salary*.10
elif salary >= 60000:
    tax_amount = salary*.5
else:
    tax_amount = salary*0

print(f"your salary is {salary}")
print(f"application of tax on your salary is {tax_amount}")
print(f"your net salary is {salary - tax_amount}")

```
