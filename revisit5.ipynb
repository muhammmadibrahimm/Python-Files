```python
def text_analyzer(txt):
    length = len(txt)
    print(f"Length is {length}")
    words = txt.split(" ")
    print(f"words are {len(words)}")
    sentence = txt.split(".")
    print(f"sentence are {len(sentence)}")
    paragrah = txt.split("\n")
    print(f"paragraph are {len(paragrah)}")
    vow = 'aeiouAEIOU'
    punc = '!#$%&\'()*+,-./:;<=>?@[\\]^_`{|}~' 
    alphabets = 0
    spaces = 0
    numbers = 0 
    lower_case = 0
    upper_case = 0
    vowels = 0
    punctuation = 0
    others = 0
    show_others = ''
    for char in txt:
        if char.isalpha():
            alphabets += 1
            if char.isupper():
                upper_case += 1
            else:
                lower_case += 1
            if char in vow:
                vowels+=1
        elif char.isspace():
            spaces+=1
        elif char.isnumeric():
            numbers+=1
        elif char in punc:
            punctuation+=1
        else:
            others+=1
            show_others+=char
    print(f""" 
        Alphabets : {alphabets}
        Upper Case : {upper_case}
        Lower Case : {lower_case}
        Spaces : {spaces}
        Numbers : {numbers}
        Vowels : {vowels}
        Punctuation : {punctuation}
        Others : {others}
        Others are : {show_others}
        """)
```

```python
text_analyzer('''Alfredo James Pacino (born April 25, 1940) is a legendary American actor celebrated for his intense, explosive, and disciplined acting style. Rising to fame in the 1970s, he became a cultural icon playing complex antiheroes, immortalized by roles like Michael Corleone in The Godfather and Tony Montana in Scarface.''')
```
